# 교차 검증(Cross Validation)

### KFold교차 검증

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# KFold
# create k num of fold set to train and test data k times & then calculate accuracy score for each time then find mean value of those accuracy scores
# .split() -> provide tuple of (train_indexes_array,test_indexes_array)
        # e.g., k = 5
        # print(list(kfold.split(features)) will return :
'''
            [
                (
                    # Train data
                    array([30, 31, ..., 149]), #except 1st batch 
                    # Test data
                    array([0,1,2,3,4,...,29]) #1st batch
                ),
                (
                    array([1, 2, ..., 149]), #except 2nd batch 
                    array([30,1,2,3,4,...,59]) #2nd batch
                )
                (
                    array([1, 2, ..., 149]), #except 3rd batch
                    array([60,1,2,3,4,...,89]) #3rd batch
                ),
                (
                    array([1, 2, ..., 149]), #except 4th batch 
                    array([90,1,2,3,4,...,119]) #4th batch
                )
                (
                    array([1, 2, ..., 119]), #except 5th batch 
                    array([120,1,2,3,4,...,149]) #5th batch
                )   
            ]
'''
from sklearn.model_selection import KFold

# Load data
iris_data = load_iris()
# Feature
#  [[5.1 3.5 1.4 0.2]
#  [4.9 3.  1.4 0.2]
#     ...
#  [4.7 3.2 1.3 0.2]]
feature = iris_data.data
# Label
# [0 0 0 ... 1 1 1 ... 2 2 2]
label = iris_data.target

# create DecisionTreeClassifier obj
dt_clf = DecisionTreeClassifier(random_state=121)

# Split data using KFold
kfold = KFold(n_splits=5)

# list of accuracy_scores
cv_accuracy = []

print('붓꽃 데이터 세트 크기:',feature.shape[0])

# num of iteration
n_iter = 0
for train_index,test_index in kfold.split(feature):
    # Feature train & test data
    X_train, X_test = feature[train_index], feature[test_index]
    # Label train & test data
    y_train, y_test = label[train_index], label[test_index]

    # Train data
    dt_clf.fit(X_train,y_train) 

    # Predict 
    pred = dt_clf.predict(X_test)

    # Calculate accuracy
    import numpy as np
    accuracy = np.round(accuracy_score(y_test,pred),4)
    cv_accuracy.append(accuracy)

    n_iter += 1

    # Display
    print('#{} 교차 검증 정확도 : {}, 학습 데이터 크기: {}, 테스트 데이터 크기: {}'.format(n_iter,accuracy,X_train.shape[0],X_test.shape[0]))
    print('#{} 검증 세트 인덱스 : {}'.format(n_iter,test_index))

print('\n평균 검증 정확도:',np.round(np.mean(cv_accuracy),4))
```

```python
# Stratified KFold
# method where if one label is imbalanced(too many or too less)
    # e.g., Fraud Cases where number of Normal Cases(0) is extremely greater than Fraud Cases(1)
        # label 1 is important to detect fraud but if we distribute with KFold there can be group with too less 1 and even no 1 included
    # uses label data info and uniformly distribute them
```

```python
# Problem with KFold
import pandas as pd
from sklearn.model_selection import KFold

# Load data
iris_data = load_iris()

# change to dataframe
#   sepal length (cm)	sepal width (cm)	petal length (cm)	petal width (cm)
# 0	             5.1	             3.5	             1.4	            0.2
# 1	             4.9	             3.0	             1.4	            0.2
# 2                ...
iris_df = pd.DataFrame(data=iris_data.data,columns=iris_data.feature_names)
iris_df['label'] = iris_data.target
# 0 50
# 1 50
# 2 50
iris_df['label'].value_counts() 

# Kfold
kfold = KFold(n_splits=3)
n_iter = 0
for train_index,test_index in kfold.split(iris_df):
    label_train =iris_df['label'].iloc[train_index]
    label_test = iris_df['label'].iloc[test_index]
    n_iter += 1
    print('교차 검증:',n_iter)
    print('학습 레이블 데이터 분포:\n',label_train.value_counts())
    print('테스트 레이블 데이터 분포:\n',label_test.value_counts())
```

```python
# Using StratifiedKFold to show that data can be distributed uniformly
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=3)
n_iter = 0
# the only difference with KFold is that skf needs label data too cuz it will distribute data according to label data
for train_index,test_index in skf.split(iris_df,iris_df['label']):
    label_train =iris_df['label'].iloc[train_index]
    label_test = iris_df['label'].iloc[test_index]
    n_iter += 1
    print('교차 검증:',n_iter)
    print('학습 레이블 데이터 분포:\n',label_train.value_counts())
    print('테스트 레이블 데이터 분포:\n',label_test.value_counts())
```

```python
# Implementing StratifiedKFold
from sklearn.model_selection import StratifiedKFold
iris_data = load_iris()
feature = iris_data.data
label = iris_data.target

dt_clf = DecisionTreeClassifier(random_state=121)
skf = StratifiedKFold(n_splits=3)
n_iter = 0
cv_accuracy = []

for train_index,test_index in skf.split(feature,label):
    X_train,X_test = feature[train_index], feature[test_index]
    y_train,y_test = label[train_index], label[test_index]

    dt_clf.fit(X_train,y_train)
    pred = dt_clf.predict(X_test)

    accuracy = np.round(accuracy_score(y_test,pred),4)
    cv_accuracy.append(accuracy)
    n_iter+=1

    # Display
    print('#{} 교차 검증 정확도 : {}, 학습 데이터 크기: {}, 테스트 데이터 크기: {}'.format(n_iter,accuracy,X_train.shape[0],X_test.shape[0]))
    print('#{} 검증 세트 인덱스 : {}'.format(n_iter,test_index))

print('\n평균 검증 정확도:',np.round(np.mean(cv_accuracy),4))
```

## 교차 검증을 보다 간편하게 - cross_val_score()

```python
from sklearn.model_selection import cross_val_score
# Processes that can be done with several steps using KFold can be done in 1 step with cross_val_score()
    # Steps:
        # Set fold set
        # Extract train_index & test_index using for loop
        # Train & Predict data repeatedly in the loop and return accuracy_score
# cross_val_score(estimator, X, y, scoring, cv, ..)
    # estimator = classifier or regressor (dt_clf)
    # X = feature dataset
    # y = label dataset
    # scoring = evaluation metrics
    # cv = number of cross validation fold
    # return accuracy_score values in array
# use StratifiedKFold 
# cross_validate() also exists -> various evaluation metrics can be used

iris_data = load_iris()
feature = iris_data.data
label = iris_data.target

dt_clf = DecisionTreeClassifier(random_state=121)

scores = cross_val_score(dt_clf,feature,label,scoring='accuracy',cv=3)
print('교차 검증별 정확도:',np.round(scores,4))
print('평균 검증 정확도:',np.round(np.mean(scores),4))
```

## GridSearchCV - 교차 검증과 최적 하이퍼 파라미터 튜닝을 한 번에

```python
# GridSearchCV
    # extract most optimal hyperparameter by inputting different hyperparameters sequentially 
# GridSearchCV(estimator, param_grid, scoring, cv, refit)
    # estimator : classifier, regressor, pipeline
    # param_grid : dictionary that contain key+list
    # scoring : evaluation metrics
    # cv : number of train/test dataset to be splitted for cross-validation
    # refit : default is True, if True find most optimal hyperparameter and train the estimator object with that hyperparameter

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

iris_data = load_iris()
feature = iris_data.data
label = iris_data.target


X_train,X_test,y_train,y_test = train_test_split(feature,label,test_size=0.2,random_state=121)

dtree = DecisionTreeClassifier()

parameters = {'max_depth':[1,2,3],'min_samples_split':[2,3]}

# 6 times of train&test
    # 3 fold sets(cv=3) * 2 hyperparameters(max-depth,min-samples_split)
# grid_dtree.cv_results_ : results of 
grid_dtree = GridSearchCV(dtree,param_grid=parameters,cv=3,refit=True)

# Train data
grid_dtree.fit(X_train,y_train)
# after training(fit()), 
    # result set of gridsearchcv is stored in .cv_results_ in dictionary form
        # {'mean_fit_time': array([0.00181341, 0.00158334, 0.00122388, 0.00110825, 0.00104499, 0.0010794 ]), ... }
    # best parameters values are stored in best_params_
    # best mean_test_score is stored in best_score_
    # when refit=True, estimator trained with most optimal parameter is stored in best_estimator_

# change into dataframe for easier display
scores_df = pd.DataFrame(data=grid_dtree.cv_results_)
# params : values of hyperparameter in each loop
# split0_test_score, split1_test_score, split3_test_score : evaluation score for each folding set (3 cuz cv=3)
# mean_test_score : mean value of split_test_scores (if cv=3, mean value of 3 evaluation scores)
# rank_test_score : rank of evaluation score
print(scores_df[['params','mean_test_score','rank_test_score','split0_test_score','split1_test_score','split2_test_score']])

# best parameters
print('GridSearchCV 최적 파라미터:',grid_dtree.best_params_)
# best score
print('GridSearchCV 최고 정확도: {0:.4f}'.format(grid_dtree.best_score_))

# best estimator
estimator = grid_dtree.best_estimator_
# Predict using best estimator
pred = estimator.predict(X_test)
# Evaluate
print('테스트 데이터 세트 정확도: {0:.4f}'.format(accuracy_score(y_test,pred)))
```

## 피처 스케일링(Feature Scaling)과 정규화(Normalization)

```python
# Feature Scaling
    # 표준화(Standardization) & 정규화(Normalization)

# Standardization
    # mean:0 & distribution:1
    # xi_new = xi - mean(x) / stdev(x)

# Normalization
    # to make two different feature have similar scale
        # e.g., 
            # distance : 1~100KM
            # money : 1~100,000,000,000Won 
            # make both to have the same unit like 0~1 
    # common normalization
        # xi_new = xi-min(x) / max(x)-min(x)
    # Vector Normalization (scikit-learn Normalizer)
        # xi_new = xi / root(xi^2+yi^2+zi^2)
```

```python
from sklearn.preprocessing import StandardScaler
# Standard Scaler
    # class for standardization
    # used in SVM, Linear Regression, Logistic Regression

iris_data = load_iris()
feature = iris_data.data
iris_df = pd.DataFrame(data=feature,columns=iris_data.feature_names)
# Before using Standard Scaler
print('Before using Standard Scaler:')
print('feature 들의 평균 값')
print(iris_df.mean())
print('feature 들의 분산 값')
print(iris_df.var())
print()

print('After using Standard Scaler:')
scaler = StandardScaler()
scaler.fit(iris_df)
# return ndarray 
iris_scaled = scaler.transform(iris_df)
# change ndarray to df
iris_scaled_df = pd.DataFrame(data=iris_scaled,columns=iris_data.feature_names)
print('feature 들의 평균 값')
print(iris_scaled_df.mean())
print('feature 들의 분산 값')
print(iris_scaled_df.var())
```

```python
from sklearn.preprocessing import MinMaxScaler
# MinMaxScaler
    # class for normalization
feature = iris_data.data
iris_df = pd.DataFrame(data=feature,columns=iris_data.feature_names)

scaler = MinMaxScaler()
scaler.fit(iris_df)
# return ndarray
iris_scaled = scaler.transform(iris_df)

# change ndarray to df
iris_scaled_df = pd.DataFrame(data=iris_scaled,columns=iris_data.feature_names)

print('feature 들의 최솟값')
print(iris_scaled_df.min())
print('feature 들의 최댓값')
print(iris_scaled_df.max())
```

```python
# Points to be aware of
# can use fit(), transform() or fit_transform()
    # fit() : basic setting (min,max etc.)
    # transform() : return data using set data with fit
# if you use fit() & transform() with Train dataset, you don't have to use fit() again on Test dataset as the scale has to be same
# Best way : perform data scaling before splitting Training & Testing data

# Train dataset = 0~10
# Test dataset = 0~5
train_array = np.arange(10).reshape(-1,1)
test_array = np.arange(5).reshape(-1,1)

# Perfoming fit on both train & test dataset
scaler = MinMaxScaler()
# Train
# min:0, max:10 -> scale: 1/10
scaler.fit(train_array)
train_scaled = scaler.transform(train_array)
print('원본 train_array 데이터:',train_array.reshape(-1))
print('Scale된 train_array 데이터:',np.round(train_scaled.reshape(-1),1))
# Test
# min:0, max:5 -> scale: 1/5
scaler.fit(test_array)
test_scaled = scaler.transform(test_array)
test_scaled = scaler.transform(test_array)
# produce same data as different data due to scale
print('원본 test_array 데이터:',test_array.reshape(-1))
print('Scale된 test_array 데이터:',np.round(test_scaled.reshape(-1),1))

print()

# Performing fit only on train dataset
scalerM = MinMaxScaler()
scalerM.fit(train_array)
train_scaled = scalerM.transform(train_array)
print('원본 train_array 데이터:',train_array.reshape(-1))
print('Scale된 train_array 데이터:',np.round(train_scaled.reshape(-1),1))
test_scaled = scalerM.transform(test_array)
print('원본 test_array 데이터:',test_array.reshape(-1))
print('Scale된 test_array 데이터:',np.round(test_scaled.reshape(-1),1))
```

```python
from sklearn.preprocessing import LabelEncoder

# Null 처리 함수
def fillna(df):
    df['Age'].fillna(df['Age'].mean(), inplace=True)
    df['Cabin'].fillna('N', inplace=True)
    df['Embarked'].fillna('N', inplace=True)
    df['Fare'].fillna(0, inplace=True)
    return df

# 머신러닝 알고리즘에 불필요한 피처 제거
def drop_features(df):
    df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
    return df

# 레이블 인코딩 수행.
def format_features(df):
    df['Cabin'] = df['Cabin'].str[:1]
    features = ['Cabin', 'Sex', 'Embarked']
    for feature in features:
        le = LabelEncoder()
        le = le.fit(df[feature])
        df[feature] = le.transform(df[feature])
    return df

# 앞에서 설정한 데이터 전처리 함수 호출
def transform_features(df):
    df = fillna(df)
    df = drop_features(df)
    df = format_features(df)
    return df
```

```python
from sklearn.model_selection import KFold

def exec_kfold(clf, folds=5):
    kfold = KFold(n_splits=folds)
    scores = []
    
    # KFold 교차 검증  
    for iter_count , (train_index, test_index) in enumerate(kfold.split(X_titanic_df)):
        X_train, X_test = X_titanic_df.values[train_index], X_titanic_df.values[test_index]
        y_train, y_test = y_titanic_df.values[train_index], y_titanic_df.values[test_index]
        
        # Classifier 학습, 예측, 정확도 계산 
        clf.fit(X_train, y_train) 
        predictions = clf.predict(X_test)
        accuracy = accuracy_score(y_test, predictions)
        scores.append(accuracy)
        print("교차 검증 {0} 정확도: {1:.4f}".format(iter_count, accuracy))     
    
    mean_score = np.mean(scores)
    print("평균 정확도: {0:.4f}".format(mean_score)) 
# exec_kfold 호출
exec_kfold(dt_clf , folds=5) 
```

## 정확도(Accuracy)`

```python
# measure similarity between real data & predicted data
# Accuracy = num of predicted data whose result is the same with real one / total num of predicted data
# Binary Classification doesn't use only Accuracy
    # e.g., In titanic survival prediction, not using algorithms and just predicting that all woman survive & all man die will give the same value as using algorithms

# Not using algorithms
# Create customized Estimator by inheriting from a BaseEstimator class
# fit() : does not do anything
# predict() : if sex feature is 1, then 0 / if not then 1
import numpy as np
from sklearn.base import BaseEstimator

class MyDummyClassifier(BaseEstimator):
    def fit(self, X , y=None):
        pass
    
    def predict(self, X):
        pred = np.zeros( ( X.shape[0], 1 ))
        for i in range (X.shape[0]) :
            if X['Sex'].iloc[i] == 1:
                pred[i] = 0
            else :
                pred[i] = 1
        
        return pred
```

```python
from sklearn.preprocessing import LabelEncoder

def fillna(df):
    df['Age'].fillna(df['Age'].mean(), inplace=True)
    df['Cabin'].fillna('N', inplace=True)
    df['Embarked'].fillna('N', inplace=True)
    df['Fare'].fillna(0, inplace=True)
    return df

def drop_features(df):
    df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
    return df

def format_features(df):
    df['Cabin'] = df['Cabin'].str[:1]
    features = ['Cabin', 'Sex', 'Embarked']
    for feature in features:
        le = LabelEncoder()
        le = le.fit(df[feature])
        df[feature] = le.transform(df[feature])
    return df

def transform_features(df):
    df = fillna(df)
    df = drop_features(df)
    df = format_features(df)
    return df
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

titanic_df = pd.read_csv('./titanic_train.csv')
y_titanic_df = titanic_df['Survived']
X_titanic_df= titanic_df.drop('Survived', axis=1)
X_titanic_df = transform_features(X_titanic_df)
X_train, X_test, y_train, y_test=train_test_split(X_titanic_df, y_titanic_df,
                                                  test_size=0.2, random_state=0)

myclf = MyDummyClassifier()
myclf.fit(X_train, y_train)

mypredictions = myclf.predict(X_test)
print('Dummy Classifier의 정확도는: {0:.4f}'.format(accuracy_score(y_test, mypredictions)))
```
