### 21. Given a DataFrame of random numeric values:
```python
df = pd.DataFrame(np.random.random(size=(5, 3))) # this is a 5x3 DataFrame of float values
```

how do you subtract the row mean from each element in the row?

````python
# 랜덤 시드 고정
np.random.seed(2025)

df = pd.DataFrame(np.random.random(size=(5, 3)))
print(df)
# to get mean of row, have to find mean of each col
row_means = df.mean(axis=1)
print(row_means)
# subtract row_mean from each element in the row
df_sub = df.sub(row_means,axis=0)
df_sub
````
